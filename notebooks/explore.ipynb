```python
"""Step 1.1b: cut off the signal peptide."""
from Bio import SeqIO
from pathlib import Path

CLEAVAGE_POS = 18  # <-- the last residue of the signal peptide from SignalP

rec = next(SeqIO.parse("data/raw/escapin_uniprot.fasta", "fasta"))
mature = rec[CLEAVAGE_POS:]
mature.id = rec.id + "_mature"
mature.description = f"signal peptide 1-{CLEAVAGE_POS} removed (SignalP 6.0)"

Path("data/processed").mkdir(parents=True, exist_ok=True)
SeqIO.write(mature, "data/processed/escapin_mature.fasta", "fasta")
print(f"Full: {len(rec)} aa, mature: {len(mature)} aa")
```
